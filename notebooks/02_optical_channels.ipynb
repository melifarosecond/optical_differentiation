```python
from svetlanna import Wavefront
from optical_differentiation.svetlanna_data import get_wavefront_datasets
from optical_differentiation.svetlanna_model import OpticalClassifier

_, test_data = get_wavefront_datasets()
model = OpticalClassifier()
model.eval()

for i in range(3):
    wavefront, label = test_data[i]
    pred = model(Wavefront(wavefront))
    print(f"label={label}, logits={pred.detach().numpy()}")
```
